### <center> Data Analysis and Cleaning

```python
print("x_train shape:", x_train.shape)
print("x_test shape:", x_test.shape)
print("y_train shape:", y_train.shape)

print("x_train dtype:", x_train.dtype)
print("y_train dtype:", y_train.dtype)
total_observations = x_train.shape[0] + x_test.shape[0]

print(f"We have a split of {x_train.shape[0]/total_observations:.2%} "
      f"for training and {x_test.shape[0]/total_observations:.2%} are for " 
      f"testing.")
```

```python
# Checking balance of the classes
unique, counts = np.unique(y_train, return_counts=True)

plt.bar(unique.astype(str), counts, color=[COLORS["blue"], COLORS["orange"]])

plt.xlabel("Class")
plt.ylabel("Number of samples")
plt.title("Class distribution in the training set")
plt.grid(False)

plt.show()
```

```python
# split the training set into training and validation sets
indices = np.random.permutation(Ntr)
split = int(0.8 * Ntr)

train_idx = indices[:split]
val_idx = indices[split:]

x_tr = x_train[train_idx]
y_tr = y_train[train_idx]

x_val = x_train[val_idx]
y_val = y_train[val_idx]

print("Training set:", x_tr.shape)
print("Validation set:", x_val.shape)
```

```python
print("NaNs in training features:", np.isnan(x_tr).sum())
print("NaNs in validation features:", np.isnan(x_val).sum())
```
